# Data Cleaning
We will clean duplicates, missing values and inconsistent text.


```python
import pandas as pd

df = pd.read_csv("chai_sales.csv")
print("Original shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
```

## Remove duplicate rows


```python
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

## Fix text values


```python
df["city"] = df["city"].str.strip().str.title()
df["payment_method"] = df["payment_method"].str.strip().str.upper()

print(df["city"].unique())
print(df["payment_method"].unique())
```

## Handle missing numerical values


```python
df["customer_rating"] = df["customer_rating"].fillna(df["customer_rating"].median())
df["temperature_c"] = df["temperature_c"].fillna(df["temperature_c"].median())
df["prep_time_min"] = df["prep_time_min"].fillna(df["prep_time_min"].median())
```

## Handle missing categorical values


```python
df["milk_type"] = df["milk_type"].fillna(df["milk_type"].mode()[0])
df["repeat_customer"] = df["repeat_customer"].fillna(df["repeat_customer"].mode()[0])

print(df.isnull().sum())
```

## Save cleaned dataset


```python
df.to_csv("cleaned_chai_sales.csv", index=False)
print("cleaned_chai_sales.csv saved")
```
