# Complete Chai Sales EDA
A small end-to-end analysis using the same dataset.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("chai_sales.csv")
print("Shape:", df.shape)
df.head()
```

## 1. Check data quality


```python
print("Duplicates:", df.duplicated().sum())
print("\nMissing values:")
print(df.isnull().sum())
```

## 2. Clean the data


```python
df = df.drop_duplicates()
df["city"] = df["city"].str.strip().str.title()
df["payment_method"] = df["payment_method"].str.strip().str.upper()

for col in ["temperature_c", "customer_rating", "prep_time_min"]:
    df[col] = df[col].fillna(df[col].median())

for col in ["milk_type", "repeat_customer"]:
    df[col] = df[col].fillna(df[col].mode()[0])
```

## 3. Overall sales numbers


```python
print("Total revenue:", round(df["total_amount"].sum(), 2))
print("Average order value:", round(df["total_amount"].mean(), 2))
print("Total cups sold:", df["quantity"].sum())
```

## 4. Best selling chai


```python
chai_orders = df.groupby("chai_type")["quantity"].sum().sort_values(ascending=False)
chai_orders
```

```python
chai_orders.plot(kind="bar", title="Cups Sold by Chai Type")
plt.ylabel("Quantity")
plt.show()
```

## 5. Revenue by city


```python
city_sales = df.groupby("city")["total_amount"].sum().sort_values(ascending=False)
city_sales.plot(kind="bar", title="Revenue by City")
plt.ylabel("Revenue")
plt.show()
```

## 6. Weekend vs weekday


```python
df["is_weekend"] = df["day"].isin(["Saturday", "Sunday"])
df.groupby("is_weekend")["total_amount"].agg(["count", "mean", "sum"])
```

## 7. Weather and chai preference


```python
pd.crosstab(df["weather"], df["chai_type"])
```

## 8. Busy hours


```python
hourly_orders = df.groupby("order_hour")["order_id"].count()
hourly_orders.plot(kind="line", marker="o", title="Orders by Hour")
plt.xlabel("Hour")
plt.ylabel("Orders")
plt.show()
```

## 9. Customer behaviour


```python
print(pd.crosstab(df["customer_type"], df["chai_type"]))
print("\nAverage spend by customer type:")
print(df.groupby("customer_type")["total_amount"].mean().sort_values(ascending=False))
```

## 10. Final observations


```python
print("Top chai:", chai_orders.index[0])
print("Highest revenue city:", city_sales.index[0])
print("Most used payment:", df["payment_method"].mode()[0])
print("Average rating:", round(df["customer_rating"].mean(), 2))
```
