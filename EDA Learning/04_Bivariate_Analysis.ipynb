# Bivariate Analysis
Compare two variables and look for relationships.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("chai_sales.csv").drop_duplicates()
```

## Quantity vs total amount


```python
df.plot.scatter(x="quantity", y="total_amount", title="Quantity vs Total Amount")
plt.show()
```

## Chai type vs average sales amount


```python
avg_sales = df.groupby("chai_type")["total_amount"].mean().sort_values()
avg_sales.plot(kind="bar", title="Average Order Value by Chai Type")
plt.ylabel("Average Amount")
plt.show()
```

## Weather vs chai type


```python
weather_chai = pd.crosstab(df["weather"], df["chai_type"])
weather_chai
```

## Repeat customer vs rating


```python
df.groupby("repeat_customer")["customer_rating"].mean()
```

## Payment method vs order mode


```python
pd.crosstab(df["payment_method"], df["order_mode"])
```
