# Univariate Analysis
Analyze one column at a time.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("chai_sales.csv").drop_duplicates()
```

## Chai type frequency


```python
df["chai_type"].value_counts()
```

```python
df["chai_type"].value_counts().plot(kind="bar", title="Orders by Chai Type")
plt.ylabel("Orders")
plt.show()
```

## Order amount distribution


```python
df["total_amount"].describe()
```

```python
df["total_amount"].plot(kind="hist", bins=20, title="Order Amount Distribution")
plt.xlabel("Total Amount")
plt.show()
```

## Quantity box plot


```python
df.boxplot(column="quantity")
plt.title("Quantity Box Plot")
plt.show()
```

## Customer ratings


```python
df["customer_rating"].value_counts().sort_index().plot(kind="bar", title="Customer Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```
