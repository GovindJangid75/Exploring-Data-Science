# Multivariate Analysis
Study more than two variables together.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("chai_sales.csv").drop_duplicates()
```

## Correlation matrix


```python
numeric_cols = [
    "order_hour", "quantity", "unit_price", "discount_percent",
    "total_amount", "temperature_c", "customer_rating", "prep_time_min"
]

corr = df[numeric_cols].corr()
corr
```

```python
fig, ax = plt.subplots(figsize=(9, 6))
img = ax.imshow(corr.values)
ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation=45, ha="right")
ax.set_yticklabels(corr.columns)
fig.colorbar(img)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

## City + chai type + sales


```python
city_chai_sales = df.pivot_table(
    values="total_amount",
    index="city",
    columns="chai_type",
    aggfunc="mean"
)
city_chai_sales
```

## Weather + customer type + amount


```python
df.groupby(["weather", "customer_type"])["total_amount"].mean().unstack()
```
