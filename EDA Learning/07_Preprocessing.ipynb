# Preprocessing
Prepare selected columns for a future ML model.


```python
import pandas as pd

df = pd.read_csv("chai_sales.csv").drop_duplicates()
```

## Fill missing values


```python
for col in ["temperature_c", "customer_rating", "prep_time_min"]:
    df[col] = df[col].fillna(df[col].median())

for col in ["milk_type", "repeat_customer"]:
    df[col] = df[col].fillna(df[col].mode()[0])
```

## Convert Yes/No


```python
df["repeat_customer"] = df["repeat_customer"].map({"Yes": 1, "No": 0})
```

## Encode categorical columns


```python
model_data = df[[
    "order_hour", "city", "chai_type", "size", "quantity",
    "unit_price", "discount_percent", "payment_method",
    "customer_type", "order_mode", "weather", "temperature_c",
    "customer_rating", "prep_time_min", "repeat_customer"
]]

model_data = pd.get_dummies(
    model_data,
    columns=[
        "city", "chai_type", "size", "payment_method",
        "customer_type", "order_mode", "weather"
    ],
    drop_first=True
)

model_data.head()
```

```python
model_data.to_csv("preprocessed_chai_sales.csv", index=False)
print("Preprocessed dataset saved")
print("Final shape:", model_data.shape)
```
