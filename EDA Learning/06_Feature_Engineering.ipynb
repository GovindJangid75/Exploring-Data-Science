# Feature Engineering
Create useful columns from existing data.


```python
import pandas as pd

df = pd.read_csv("chai_sales.csv").drop_duplicates()
df["order_date"] = pd.to_datetime(df["order_date"])
```

## Date based features


```python
df["month"] = df["order_date"].dt.month
df["week_number"] = df["order_date"].dt.isocalendar().week
df["is_weekend"] = df["day"].isin(["Saturday", "Sunday"]).astype(int)
```

## Time based feature


```python
def time_slot(hour):
    if hour < 11:
        return "Morning"
    elif hour < 16:
        return "Afternoon"
    elif hour < 19:
        return "Evening"
    return "Night"

df["time_slot"] = df["order_hour"].apply(time_slot)
```

## Customer and sales features


```python
df["repeat_customer_flag"] = df["repeat_customer"].map({"Yes": 1, "No": 0})
df["discount_amount"] = (
    df["unit_price"] * df["quantity"] * df["discount_percent"] / 100
)

df[[
    "order_date", "day", "month", "is_weekend", "time_slot",
    "discount_amount", "repeat_customer_flag"
]].head()
```

```python
df.to_csv("feature_engineered_chai_sales.csv", index=False)
print("Feature engineered file saved")
```
