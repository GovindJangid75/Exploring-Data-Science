# Seaborn Basics and Core Plots


## Import Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Set Theme
`set_theme()` Seaborn ka default visual theme apply karta hai or ise we can chnag e


```python
sns.set_theme()
```

## Built-in Datasets
built-in datasets hote hia isme


```python
print(sns.get_dataset_names())
```

## Load Dataset


```python
tips = sns.load_dataset("tips")

print(type(tips))
print(tips.head())
```

## Scatter Plot
relation dikata hai jo humari kismat mein kha hai

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)

plt.title("Total Bill vs Tip")
plt.show()
```

## Scatter with More Categories
`hue`, `style` aur `size` extra information graph addition ke liye .


```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    size="size"
)

plt.show()
```

## Relplot
`relplot()` figure-level relational plot 

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker"
)

plt.show()
```

## Line Plot
Trend ya ordered data ko show karne ke liye line plot


```python
flights = sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)

plt.title("Passengers by Year")
plt.show()
```

## Line Plot with Hue
Category-wise separate lines ke liye `hue` use kr skte hai vaise toh


```python
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar=None
)

plt.title("Average Bill by Day")
plt.show()
```

## Bar Plot
numerical value compare as per catag.


```python
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)

plt.title("Average Tip by Day")
plt.show()
```

## Count Plot
rows count


```python
sns.countplot(
    data=tips,
    x="day",
    hue="sex"
)

plt.title("Customers by Day")
plt.show()
```

## Box Plot

```python
sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)

plt.title("Tip Distribution by Day")
plt.show()
```

## Violin Plot

```python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.title("Bill Distribution")
plt.show()
```

## Strip Plot
Categorical groups ke individual data points we can show by this


```python
sns.stripplot(
    data=tips,
    x="day",
    y="tip",
    hue="time",
    jitter=True
)

plt.show()
```

## Histogram
fir se aa gya tu reeee


```python
penguins = sns.load_dataset("penguins")

sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=20
)

plt.title("Penguin Body Mass Distribution")
plt.show()
```

## Histogram with KDE
`kde=True` cureve density


```python
sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    kde=True
)

plt.show()
```

## KDE Plot

```python
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="sex",
    fill=True
)

plt.show()
```

## Basic Styling
Graph sundar , data galat


```python
sns.set_theme(style="whitegrid")

sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Average Bill")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```
