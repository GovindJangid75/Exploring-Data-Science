# Seaborn Advanced Plots and Practice


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme()

tips = sns.load_dataset("tips")
flights = sns.load_dataset("flights")
penguins = sns.load_dataset("penguins")
```

## Heatmap
Matrix/pivot data ko color intensity 


```python
flights_pivot = flights.pivot(
    index="month",
    columns="year",
    values="passengers"
)

sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)

plt.title("Passengers Heatmap")
plt.tight_layout()
plt.show()
```

## Correlation Heatmap


```python
corr = tips.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

## Pair Plot
pair toh relationship mein hota hai so tu whi ja yha baar baar dikh jata hai ree


```python
sns.pairplot(
    penguins,
    hue="species"
)

plt.show()
```

## Joint Plot
Relationship + individual distributions in one fig


```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.show()
```

## Catplot
`catplot()` figure-level categorical plotting function 


```python
sns.catplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex",
    kind="bar"
)

plt.show()
```

## Displot
`displot()` figure-level distribution plot


```python
sns.displot(
    data=tips,
    x="total_bill",
    hue="sex",
    kde=True
)

plt.show()
```

## Relplot with Columns

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    col="time"
)

plt.show()
```

## FacetGrid
Custom faceted visualizations keliye we can use FacetGrid 


```python
g = sns.FacetGrid(
    tips,
    col="time",
    row="sex"
)

g.map_dataframe(
    sns.scatterplot,
    x="total_bill",
    y="tip"
)

plt.show()
```

## Seaborn with Matplotlib Axes

```python
fig, ax = plt.subplots(figsize=(7, 4))

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar=None,
    ax=ax
)

ax.set_title("Avg Bill Value For Each Day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
plt.show()
```

## Multiple Subplots
different Seaborn plots can be combine


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    ax=axes[0]
)
axes[0].set_title("Tip Box Plot")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time",
    ax=axes[1]
)
axes[1].set_title("Bill vs Tip")

plt.tight_layout()
plt.show()
```

```python
sns.swarmplot(
    data=tips,
    x="day",
    y="tip",
    size=3
)

plt.title("Tips by Day")
plt.show()
```
