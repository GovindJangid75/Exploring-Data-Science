# Joining, Random, Linear Algebra and Practice


```python
import numpy as np
```

## Concatenate
Arrays ko existing axis par join karta hai.


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate((a, b)))
```

## Stacking
`vstack`, `hstack` aur `stack` useful hain.


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.vstack((a, b)))
print(np.hstack((a, b)))
print(np.stack((a, b), axis=0))
```

## Splitting
Array ko multiple parts me split.


```python
arr = np.array([1, 2, 3, 4, 5, 6])

print(np.split(arr, 3))
```

## Random Numbers
Modern NumPy me `default_rng()` recommended approach hai.


```python
rng = np.random.default_rng(42)

print(rng.random(5))
print(rng.integers(1, 10, size=5))
print(rng.normal(0, 1, size=5))
```

## Random 2D Array
Shape dekar random matrix bana sakte hain.


```python
rng = np.random.default_rng(10)

random_matrix = rng.integers(1, 100, size=(3, 4))
print(random_matrix)
```

## Shuffle and Choice
Random selection ke kaam aata hai.


```python
rng = np.random.default_rng(5)

arr = np.array([10, 20, 30, 40, 50])

rng.shuffle(arr)
print(arr)

print(rng.choice(arr, size=3))
```

## Dot Product
Vectors ka dot product.


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.dot(a, b))
```

## Matrix Multiplication
`@` ya `np.matmul()` use kar sakte hain.


```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(a @ b)
print(np.matmul(a, b))
```

## Linear Algebra
`np.linalg` me matrix related functions hain.


```python
matrix = np.array([[4, 7], [2, 6]])

print("Determinant:", np.linalg.det(matrix))
print("Inverse:")
print(np.linalg.inv(matrix))
```

## Solve Equations
`np.linalg.solve()` linear equations solve karta hai.


```python
# 2x + y = 5
# x - y = 1

A = np.array([[2, 1], [1, -1]])
B = np.array([5, 1])

solution = np.linalg.solve(A, B)
print(solution)
```
