# Pandas Fundamentals


## Import Pandas
Pandas ko usually `pd` or kisi name se kr skte hai


```python
import pandas as pd
```

## Series
Series -> 1D labeled data.


```python
chai = pd.Series(["Masala", "Ginger", "Elaichi", "Lemon"])
print(chai)

prices = pd.Series([20, 30, 25, 15], index=["masala", "ginger", "elaichi", "lemon"])
print(prices)
print(prices["ginger"])
```

## DataFrame
DataFrame -> rows + columns wala tabular data.


```python
data = {
    "name": ["Govind", "Aman",  "Gourav", "Ayush", "Chaman"],
    "age": [20, 21, 19, 22, 20],
    "city": ["Jaipur", "Delhi", "Jaipur", "Mumbai", "Delhi"],
    "marks": [85, 92, 76, 88, 69]
}

df = pd.DataFrame(data)
df
```

## Basic Inspection
`head`, `tail`, `shape`, `columns`, `dtypes`, `info` basic checking ke liye.


```python
print(df.head())
print(df.tail(2))
print(df.shape)
print(df.columns)
print(df.dtypes)
df.info()
```

## describe()
Numerical columns ka quick summary.


```python
df.describe()
```

## Index
DataFrame ka index check/change krna vgera


```python
print(df.index)

temp = df.set_index("name")
print(temp)

temp = temp.reset_index()
print(temp)
```

## Create from List of Dictionaries
List of dictionaries bnaooo



```python
students = [
    {"name": "Govind", "marks": 80},
    {"name": "Ayush", "marks": 91},
    {"name": "Gourav", "marks": 73}
]

pd.DataFrame(students)
```
