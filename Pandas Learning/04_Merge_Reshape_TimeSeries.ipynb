# Merge, Reshape and Time Series


## Setup
Different DataFrames ko combine karna Pandas ka important part hai.


```python
import pandas as pd

students = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Govind", "pta nhi", "garu"]
})

marks = pd.DataFrame({
    "id": [1, 2, 4],
    "marks": [80, 92, 75]
})
```

## merge - inner
Common keys wale records.


```python
pd.merge(students, marks, on="id", how="inner")
```

## merge - left/right/outer
Join type requirement ke according choose karo.


```python
print(pd.merge(students, marks, on="id", how="left"))
print(pd.merge(students, marks, on="id", how="right"))
print(pd.merge(students, marks, on="id", how="outer"))
```

## concat
DataFrames ko row-wise ya column-wise combine.


```python
a = pd.DataFrame({"name": ["Govind", "pta nhi"]})
b = pd.DataFrame({"name": ["govind", "garru"]})

print(pd.concat([a, b], ignore_index=True))

left = pd.DataFrame({"name": ["Govind", "Ashi"]})
right = pd.DataFrame({"marks": [80, 90]})
print(pd.concat([left, right], axis=1))
```

## Pivot
Long data ko wide format me.


```python
sales = pd.DataFrame({
    "city": ["Jaipur", "Jaipur", "Delhi", "Delhi"],
    "product": ["Tea", "Coffee", "Tea", "Coffee"],
    "sales": [100, 150, 120, 180]
})

sales.pivot(index="city", columns="product", values="sales")
```

## Melt
Wide data ko long format me.


```python
wide = pd.DataFrame({
    "name": ["Aman", "Riya"],
    "Math": [80, 90],
    "Science": [85, 88]
})

pd.melt(
    wide,
    id_vars="name",
    var_name="subject",
    value_name="marks"
)
```

## DateTime
`to_datetime` se date column proper datetime me convert.


```python
orders = pd.DataFrame({
    "date": ["2026-01-10", "2026-02-15", "2026-03-20"],
    "sales": [100, 150, 220]
})

orders["date"] = pd.to_datetime(orders["date"])
print(orders.dtypes)
```

## Date Parts
Datetime se year/month/day nikal sakte hain.


```python
orders["year"] = orders["date"].dt.year
orders["month"] = orders["date"].dt.month
orders["day"] = orders["date"].dt.day

orders
```

## Datetime Index and Resample
Time-series data ko period-wise summarize.


```python
daily = pd.DataFrame({
    "date": pd.date_range("2026-01-01", periods=10, freq="D"),
    "sales": [10, 15, 12, 20, 25, 18, 30, 22, 35, 40]
})

daily = daily.set_index("date")
print(daily.resample("5D")["sales"].sum())
```

## Rolling
Moving average jaise calculations.


```python
daily["rolling_3"] = daily["sales"].rolling(3).mean()
daily
```
