# File I/O, Advanced Pandas and Practice


## Setup
Final notebook me file handling + useful advanced operations.


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Govind", "Garru", "gourav", "dilip"],
    "department": ["CSE", "CSE", "IT", "IT"],
    "marks": [82, 94, 73, 88],
    "email": [
        " Govind@gmail.com ",
        "GARRU@GMAIL.COM",
        "GOURAV@yahoo.com",
        "DILIP@gmail.com"
    ]
})

df
```

## String Methods
`str` accessor text cleaning ke liye.


```python
df["email"] = df["email"].str.strip().str.lower()
print(df["email"])

print(df["email"].str.contains("gmail"))
print(df["name"].str.upper())
```

## Conditional Column
`np.where` simple conditions ke liye useful.


```python
df["result"] = np.where(df["marks"] >= 75, "Pass", "Improve")
df
```

## cut
Numerical values ko categories/bins me divide.


```python
df["grade"] = pd.cut(
    df["marks"],
    bins=[0, 60, 75, 90, 100],
    labels=["D", "C", "B", "A"]
)

df
```

## Rank
Values ko rank dena.


```python
df["rank"] = df["marks"].rank(ascending=False)
df
```

## CSV Write and Read
`to_csv` save, `read_csv` read.


```python
df.to_csv("students.csv", index=False)

loaded = pd.read_csv("students.csv")
loaded
```
