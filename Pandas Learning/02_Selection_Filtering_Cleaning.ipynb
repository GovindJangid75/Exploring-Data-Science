# Selection, Filtering and Cleaning


## Setup
Practice DataFrame.


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Govind", "Ayush", "Gourav", "Garvit", "chaman", "Aman"],
    "age": [20, 21, 19, 22, None, 20],
    "city": ["Jaipur", "Delhi", "Jaipur", None, "Delhi", "Jaipur"],
    "marks": [85, 92, 76, 88, 69, 85]
})

df
```

## Selecting Columns
Single column -> Series, multiple columns -> DataFrame.


```python
print(df["name"])
print(df[["name", "marks"]])
```

## loc
`loc` -> labels/names based selection.


```python
print(df.loc[0])
print(df.loc[0, "name"])
print(df.loc[0:2, ["name", "marks"]])
```

## iloc
`iloc` -> position based selection.


```python
print(df.iloc[0])
print(df.iloc[0:3, 0:2])
```

## Basic Filtering
Condition laga ke rows filter.


```python
print(df[df["marks"] > 80])
print(df[df["city"] == "Jaipur"])
```

## Multiple Conditions
`&` = AND, `|` = OR. Conditions ke around brackets lagao.


```python
print(df[(df["marks"] > 80) & (df["age"] >= 20)])
print(df[(df["city"] == "Jaipur") | (df["city"] == "Delhi")])
```

## isin, between, query
Useful filtering shortcuts.


```python
print(df[df["city"].isin(["Jaipur", "Delhi"])])
print(df[df["marks"].between(80, 90)])
print(df.query("marks > 80"))
```

## Missing Values
`isna()` se missing values check.


```python
print(df.isna())
print(df.isna().sum())
```

## fillna
Missing data fill karne ke liye.


```python
clean = df.copy()
clean["age"] = clean["age"].fillna(clean["age"].mean())
clean["city"] = clean["city"].fillna("Unknown")
clean
```

## dropna
Missing values wali rows remove.


```python
print(df.dropna())
print(df.dropna(subset=["city"]))
```

## Duplicates
Duplicate rows find/remove.


```python
print(df.duplicated())
print(df.drop_duplicates())
```

## Replace
Values replace karna.


```python
changed = df.replace({"Jaipur": "JPR", "Delhi": "DEL"})
changed
```

## Rename
Columns rename.


```python
df.rename(columns={"name": "student_name", "marks": "score"})
```
