# Data Manipulation and GroupBy



sample sales data of chai


```python
import pandas as pd

df = pd.DataFrame({
    "product": ["Tea", "Coffee", "Tea", "Coffee", "Juice", "Tea"],
    "city": ["Jaipur", "Jaipur", "Delhi", "Delhi", "Jaipur", "Delhi"],
    "price": [20, 50, 25, 55, 40, 30],
    "quantity": [3, 2, 4, 1, 2, 5]
})

df
```

## Add Columns
Existing columns se new column.


```python
df["total"] = df["price"] * df["quantity"]
df
```

## Update Values
`loc` se selected value update.


```python
df.loc[0, "price"] = 22
df
```

## Drop Columns and Rows
`drop` se rows/columns remove.


```python
print(df.drop(columns=["quantity"]))
print(df.drop(index=0))
```

## Sorting
`sort_values` se sorting.


```python
print(df.sort_values("price"))
print(df.sort_values("total", ascending=False))
print(df.sort_values(["city", "price"], ascending=[True, False]))
```

## Basic Aggregations
Common calculations.


```python
print("Mean:", df["price"].mean())
print("Max:", df["price"].max())
print("Min:", df["price"].min())
print("Sum:", df["total"].sum())
print("Count:", df["product"].count())
```

## value_counts and unique
Category frequency aur unique values.


```python
print(df["product"].value_counts())
print(df["city"].unique())
print(df["city"].nunique())
```

## GroupBy
Same category ke records ko group karke calculation.


```python
print(df.groupby("product")["total"].sum())
print(df.groupby("city")["price"].mean())
```

## agg
Ek saath multiple aggregations.


```python
df.groupby("product").agg(
    avg_price=("price", "mean"),
    total_sales=("total", "sum"),
    max_quantity=("quantity", "max")
)
```

## transform
Group result ko original rows ke size me return karta hai.


```python
df["city_avg_price"] = df.groupby("city")["price"].transform("mean")
df
```

## apply and map
Custom logic apply kar sakte hain.


```python
df["status"] = df["total"].apply(
    lambda x: "Badi party ab toh paisa hi paisa hoga " if x >= 100 else "Normal order ,ye garib "
)

df["city_code"] = df["city"].map({
    "Jaipur": "JPR",
    "Delhi": "DEL"
})

df
```

## Pivot Table
Summary table banane ke liye.


```python
pd.pivot_table(
    df,
    values="total",
    index="product",
    columns="city",
    aggfunc="sum",
    fill_value=0
)
```
