# Matplotlib Basics and Line Plots


## Import Matplotlib



```python
import matplotlib.pyplot as plt
```

## First Plot
`plot()` basic line graph banata hai.


```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.show()
```

## Title and Labels


```python
plt.plot(x, y)

plt.title("Sales Report")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
```

## Marker and Line Style
Marker points ko aur linestyle line ko apne according kr skte hai



```python
plt.plot(
    x, y,
    marker="o",
    linestyle="--",
    linewidth=2
)

plt.title("Sales")
plt.show()
```

## Multiple Lines
Same graph par multiple datasets compare 


```python
days = [1, 2, 3, 4, 5]
tea = [20, 25, 30, 28, 35]
coffee = [15, 18, 22, 25, 27]

plt.plot(days, tea, marker="o", label="Tea")
plt.plot(days, coffee, marker="s", label="Coffee")

plt.xlabel("Day")
plt.ylabel("Cups")
plt.title("Tea vs Coffee")
plt.legend()
plt.show()
```

## Grid
readability increase krna


```python
plt.plot(days, tea, marker="o")
plt.grid(True)
plt.show()
```

## Axis Limits
`xlim` aur `ylim` 


```python
plt.plot(days, tea)

plt.xlim(1, 5)
plt.ylim(0, 40)

plt.show()
```

## Figure Size
`figsize` width and height inches


```python
plt.figure(figsize=(8, 4))
plt.plot(days, tea, marker="o")
plt.title("Tea Sales")
plt.show()
```

## Ticks
Axis par displayed values modified


```python
plt.plot(days, tea, marker="o")

plt.xticks(days)
plt.yticks([0, 10, 20, 30, 40])

plt.show()
```

## Save Figure
`savefig()` graph ko image file mein save krne ke liye use hota 


```python
plt.plot(days, tea, marker="o")
plt.title("Tea Sales")

plt.savefig("tea_sales.png", dpi=200, bbox_inches="tight")
plt.show()
```
