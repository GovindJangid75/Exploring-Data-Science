# Bar, Scatter, Histogram and Pie Charts


## Bar Chart
Categories Dataset kocompare karne ke liye


```python
import matplotlib.pyplot as plt

products = ["Tea", "Coffee", "Juice", "Milk"]
sales = [120, 90, 70, 100]

plt.bar(products, sales)
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

## Horizontal Bar
`barh()` 


```python
plt.barh(products, sales)
plt.title("Product Sales")
plt.xlabel("Sales")
plt.show()
```

## Grouped Bar Chart
Do datasets ek sath compare krna


```python
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr"]
tea = [40, 55, 60, 70]
coffee = [35, 50, 45, 65]

x = np.arange(len(months))
width = 0.35

plt.bar(x - width/2, tea, width, label="Tea")
plt.bar(x + width/2, coffee, width, label="Coffee")

plt.xticks(x, months)
plt.legend()
plt.title("Monthly Sales")
plt.show()
```

## Stacked Bar Chart
`bottom` use karke bars stack kar sakte hai


```python
plt.bar(months, tea, label="Tea")
plt.bar(months, coffee, bottom=tea, label="Coffee")

plt.title("Stacked Sales")
plt.legend()
plt.show()
```

## Scatter Plot
relationship show krne ke liye


```python
study_hours = [1, 2, 3, 4, 5, 6]
marks = [45, 52, 60, 68, 78, 88]

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

## Scatter Size and Alpha
`s` size aur `alpha` transparency control


```python
sizes = [40, 60, 80, 100, 120, 140]

plt.scatter(study_hours, marks, s=sizes, alpha=0.6)
plt.grid(True)
plt.show()
```

## Histogram
Continuous data ki distribution vgera ke use kr skte hai


```python
ages = [18, 19, 20, 20, 21, 21, 21, 22, 23, 24, 24, 25, 26, 27]

plt.hist(ages, bins=5, edgecolor="black")

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

## Multiple Histograms


```python
legit_transactions = [10, 20, 25, 30, 35, 40, 50, 60, 75, 90]
fraud_transactions = [70, 80, 90, 100, 120, 150, 180, 200]

plt.hist(
    [legit_transactions, fraud_transactions],
    bins=6,
    label=["Legit", "Fraud"],
    alpha=0.7
)

plt.title("Transaction Distribution")
plt.legend()
plt.show()
```

## Pie Chart
Parts 


```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explode = [0, 0, 0, 0.1, 0]

plt.pie(
    amounts,
    labels=expenses,
    autopct="%1.1f%%",
    explode=explode,
    startangle=180,
    shadow=True,
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "linestyle": "--"
    }
)

plt.title("Company Expenses")
plt.show()
```

## Box Plot
Median, spread aur possible outliers 


```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False)
plt.grid(True)
plt.title("Box Plot")
plt.show()
```
