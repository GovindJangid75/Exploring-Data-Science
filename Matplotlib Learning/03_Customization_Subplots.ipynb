# Customization and Subplots


## Basic Customization
Title, labels, legend, marker, linewidth .


```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [5, 9, 7, 12, 15]

plt.plot(x, y, marker="o", linestyle="--", linewidth=2, label="Sales")

plt.title("Sales Trend")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

## Text on Graph
`text()` graph ke specific point par txt add krne ke use krte hai


```python
plt.plot(x, y, marker="o")

plt.text(3, 7, "Low point")
plt.title("Sales")
plt.show()
```

## Annotation


```python
plt.plot(x, y, marker="o")

plt.annotate(
    "Highest",
    xy=(5, 15),
    xytext=(3.5, 16),
    arrowprops={"arrowstyle": "->"}
)

plt.show()
```

## Subplot - Basic
`subplot(rows, columns, position)` 


```python
x = [1, 2, 3, 4, 5]

plt.subplot(1, 2, 1)
plt.plot(x, [i**2 for i in x])
plt.title("Square")

plt.subplot(1, 2, 2)
plt.plot(x, [i**3 for i in x])
plt.title("Cube")

plt.tight_layout()
plt.show()
```

## Four Subplots


```python
x = np.arange(1, 11)

y1 = np.sqrt(x)
y2 = x ** 2
y3 = x ** 3
y4 = x * 10

plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title("Square Root")

plt.subplot(2, 2, 2)
plt.plot(x, y2)
plt.title("Square")

plt.subplot(2, 2, 3)
plt.plot(x, y3)
plt.title("Cube")

plt.subplot(2, 2, 4)
plt.plot(x, y4)
plt.title("x * 10")

plt.tight_layout()
plt.show()
```

## Object Oriented Approach
`fig, ax = plt.subplots()` 


```python
fig, ax = plt.subplots()

ax.plot(x, y2, marker="o")
ax.set_title("Square")
ax.set_xlabel("X")
ax.set_ylabel("X Square")
ax.grid(True)

plt.show()
```

## Multiple Axes
`axes` array se multiple plots keliye


```python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].plot(x, y1)
axes[0, 0].set_title("Square Root")

axes[0, 1].plot(x, y2)
axes[0, 1].set_title("Square")

axes[1, 0].plot(x, y3)
axes[1, 0].set_title("Cube")

axes[1, 1].plot(x, y4)
axes[1, 1].set_title("x * 10")

plt.tight_layout()
plt.show()
```

## Style

```python
print(plt.style.available[:10])

with plt.style.context("ggplot"):
    plt.plot(x, y2)
    plt.title("Using a Style")
    plt.show()
```

## Fill Between



```python
x = np.arange(1, 8)
sales = np.array([10, 15, 13, 20, 18, 25, 30])

plt.plot(x, sales)
plt.fill_between(x, sales, alpha=0.3)

plt.title("Weekly Sales")
plt.show()
```
