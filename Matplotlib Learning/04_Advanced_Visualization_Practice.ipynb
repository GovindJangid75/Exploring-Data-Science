# Advanced Visualization and Practice


## Step Plot



```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.step(x, y, where="mid")
plt.title("Step Plot")
plt.show()
```

## Stem Plot
Individual values ko baseline se stems   form 


```python
plt.stem(x, y)
plt.title("Stem Plot")
plt.show()
```

## Error Bar
Measurements ke uncertainty/error range


```python
values = [10, 15, 13, 18, 20]
errors = [1, 2, 1, 3, 2]

plt.errorbar(x, values, yerr=errors, marker="o", capsize=5)
plt.title("Measurements with Error")
plt.show()
```

## Time Series Plot
Dates ke saath line graph common analytics


```python
import pandas as pd

dates = pd.date_range("2026-01-01", periods=7)
sales = [100, 120, 115, 140, 160, 155, 180]

plt.figure(figsize=(8, 4))
plt.plot(dates, sales, marker="o")

plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## imshow
`imshow()` matrix/image-like data display krwane ke liue


```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

plt.imshow(matrix)
plt.colorbar()
plt.title("Matrix Visualization")
plt.show()
```

## Pandas + Matplotlib


```python
df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [120, 150, 135, 180, 210]
})

plt.plot(df["month"], df["sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

## Mini Dashboard and project


```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
sales = [12, 40, 25, 18]
prices = [60000, 800, 1500, 12000]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(products, sales)
axes[0].set_title("Units Sold")
axes[0].tick_params(axis="x", rotation=30)

axes[1].scatter(prices, sales, s=100)
axes[1].set_title("Price vs Sales")
axes[1].set_xlabel("Price")
axes[1].set_ylabel("Units Sold")

plt.tight_layout()
plt.show()
```
